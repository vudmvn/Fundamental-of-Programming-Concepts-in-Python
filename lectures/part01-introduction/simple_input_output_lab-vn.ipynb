## 8. Định dạng đầu ra với f-string

F-string giúp kết hợp văn bản, biến và biểu thức trong cùng một chuỗi theo cách rõ ràng và thuận tiện.

````python
name = "An"
age = 18

print(f"My name is {name}.")
print(f"I am {age} years old.")
````

## 1. Core Concepts to Remember

### Output

The `print()` function is used to display data on the screen.

```python
print("Hello")
```

### Input

The `input()` function receives data entered by the user from the keyboard.

```python
name = input("Name: ")
```

The value returned by `input()` always has type `str`.

When arithmetic calculations are required, convert the input to an appropriate numeric type:

```python
age = int(input("Age: "))
price = float(input("Price: "))
```

A basic input/output program is typically organized according to the following model:

```text
Input → Process → Output
```

## 7. The Input → Process → Output Model

1. **Input:** receive input data;

````python
length = float(input("Length: "))
width = float(input("Width: "))

area = length * width

print("Area:", area)
````

## 8. Formatting Output with f-Strings

F-strings make it clear and convenient to combine text, variables, and expressions in the same string.


````python
name = "An"
age = 18

print(f"My name is {name}.")
print(f"I am {age} years old.")
````

### Using Expressions Inside `{}`

In addition to variable names, the content inside `{}` can be any valid Python expression.


````python
x = 10
y = 5
print(f"{x} + {y} = {x + y}")
````

### Formatting Floating-Point Numbers

The `.2f` format displays a floating-point number with two digits after the decimal point.


````python
price = 19.5678
print(f"Price: {price:.2f}")
````

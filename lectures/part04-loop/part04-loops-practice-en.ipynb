# 4.1 The `while` Loop

## Concept Review

A `while` loop repeats a block **while its condition is true**.

```python
while condition:
    statements
```

A typical loop needs three things:

1. **Initialization** — establish the starting state.
2. **Condition** — decide whether repetition continues.
3. **Update** — change the loop state so the condition can eventually become false.

## Worked Example 4.1A — Doubling an Investment

````python
RATE = 5.0
INITIAL_BALANCE = 10000.0
TARGET = 2 * INITIAL_BALANCE

balance = INITIAL_BALANCE
year = 0

while balance < TARGET:
    year = year + 1
    interest = balance * RATE / 100
    balance = balance + interest

print("The investment doubled after", year, "years.")
print("Final balance: %.2f" % balance)
````

### Think before running

What changes if:

- `RATE = 10.0`?
- `TARGET = 3 * INITIAL_BALANCE`?
- the condition becomes `balance <= TARGET`?

## Worked Example 4.1B — Counter-Controlled `while` Loop

````python
counter = 1

while counter <= 5:
    print(counter)
    counter = counter + 1
````

## Common Errors: Loop Conditions

### Error 1 — Thinking “Are we there yet?”

The loop condition says **when the loop is allowed to continue**, not when it should stop.

For example:

```python
while balance < TARGET:
```

means:

> Keep repeating while the target has **not yet** been reached.

### Error 2 — Infinite loops

A loop can run forever when the control variable never changes appropriately.

### Error 3 — Off-by-one errors

Small differences such as `<` versus `<=` often change the number of iterations.

# 4.3 Application: Processing Sentinel Values

## Concept Review

A **sentinel** is a special value that marks the end of an input sequence but is **not part of the data**.

Example:

- salaries are non-negative;
- `-1` can therefore be used to mean “stop”.

Two common patterns are:

1. initialize the input variable before the loop;
2. use a **priming read** before the loop and read again at the bottom.

## Worked Example 4.3A — Salary Average with Sentinel

````python
# Sample data are used instead of keyboard input so the example can run automatically.
values = [10000, 10000, 40000, -1]

total = 0.0
count = 0
index = 0
salary = values[index]

while salary >= 0:
    total = total + salary
    count = count + 1

    index = index + 1
    salary = values[index]

if count > 0:
    average = total / count
    print("Average salary is", average)
else:
    print("No data was entered.")
````

## Worked Example 4.3B — Priming Read Pattern

````python
values = [12, 8, 20, -1]

index = 0
value = values[index]      # priming read
total = 0

while value >= 0:
    total = total + value

    index = index + 1
    value = values[index]  # next read

print("Total:", total)
````

## 4.5.4 Maximum and Minimum — Worked Example

````python
values = [18, 7, 25, 11, 20]

largest = values[0]
smallest = values[0]

for value in values[1:]:
    if value > largest:
        largest = value

    if value < smallest:
        smallest = value

print("Largest:", largest)
print("Smallest:", smallest)
````

### Why not initialize `largest = 0`?

If all inputs are negative, `0` is not an actual input and would incorrectly remain the largest value.

A robust beginner pattern is:

> initialize from the first real data value.

## Example — Processing Grades with Several Loop Patterns

````python
grades = [80, 45, 92, 68, 55]

num_passing = 0
num_failing = 0
total = 0
count = 0
minimum = grades[0]
maximum = grades[0]

for grade in grades:
    if grade >= 50:
        num_passing = num_passing + 1
    else:
        num_failing = num_failing + 1

    if grade < minimum:
        minimum = grade

    if grade > maximum:
        maximum = grade

    total = total + grade
    count = count + 1

average = total / count

print("Average:", average)
print("Passing:", num_passing)
print("Failing:", num_failing)
print("Maximum:", maximum)
print("Minimum:", minimum)
````

````python
print("range(6):")
for i in range(6):
    print(i, end=" ")

print("\n\nrange(10, 16):")
for i in range(10, 16):
    print(i, end=" ")

print("\n\nrange(0, 9, 2):")
for i in range(0, 9, 2):
    print(i, end=" ")

print("\n\nrange(5, 0, -1):")
for i in range(5, 0, -1):
    print(i, end=" ")
````

### Key rule

The ending value of `range(start, stop)` is **not included**.

Therefore:

```python
range(a, b)
```

contains `b - a` values when the step is `1`.

# HOW TO 4.1 — Writing a Loop

A practical design process:

1. Decide what work belongs inside the loop.
2. Specify the stopping condition or repetition count.
3. Decide whether the loop is:
   - **count-controlled**, or
   - **event-controlled**.
4. Initialize the required variables.
5. Decide what happens after the loop.
6. Hand-trace a small case.
7. Implement the code.

## Worked Design Example — Highest Monthly Value

````python
temperatures = [22.6, 36.6, 44.5, 24.2]

highest_value = temperatures[0]
highest_month = 1

for current_month in range(2, len(temperatures) + 1):
    next_value = temperatures[current_month - 1]

    if next_value > highest_value:
        highest_value = next_value
        highest_month = current_month

print("Highest month:", highest_month)
print("Highest temperature:", highest_value)
````

# 4.7 Nested Loops

## Concept Review

A loop inside another loop is a **nested loop**.

Typical uses:

- rows and columns;
- tables;
- repeated groups of repeated data;
- text patterns.

If the outer loop executes `m` times and the inner loop executes `n` times for every outer iteration, the inner body executes:

$$
m \times n
$$

times.

## 4.7.1 Worked Example — Table of Powers

````python
NMAX = 4
XMAX = 5

for x in range(1, XMAX + 1):
    for n in range(1, NMAX + 1):
        print("%8.0f" % (x ** n), end="")
    print()
````

## 4.7.2 Execution Order

````python
for i in range(3):
    for j in range(1, 4):
        print("i =", i, "j =", j)
````

For each single value of `i`, the **entire inner loop** runs through all values of `j`.

## 4.7.3 Pattern Examples

````python
# Rectangle
for i in range(3):
    for j in range(4):
        print("*", end="")
    print()
````

````python
# Triangle
for i in range(4):
    for j in range(i + 1):
        print("*", end="")
    print()
````

## 4.8.3 Finding the First Match

````python
text = "abc7de9"

found = False
position = 0

while not found and position < len(text):
    if text[position].isdigit():
        found = True
    else:
        position = position + 1

if found:
    print("First digit occurs at position", position)
else:
    print("No digit found")
````

## 4.8.4 Finding the Last Match

````python
text = "abc7de9x"

found = False
position = len(text) - 1

while not found and position >= 0:
    if text[position].isdigit():
        found = True
    else:
        position = position - 1

if found:
    print("Last digit occurs at position", position)
else:
    print("No digit found")
````

## 4.8.5 Validating a String

````python
phone = "(123)456-7890"

valid = len(phone) == 13
position = 0

while valid and position < len(phone):
    if position == 0:
        valid = phone[position] == "("
    elif position == 4:
        valid = phone[position] == ")"
    elif position == 8:
        valid = phone[position] == "-"
    else:
        valid = phone[position].isdigit()

    position = position + 1

print("Valid format:", valid)
````

````python
user_input = "4123-5678-9012-3450"

credit_card_number = ""

for char in user_input:
    if char != " " and char != "-":
        credit_card_number = credit_card_number + char

print(credit_card_number)
````

## 4.9.4 Monte Carlo Method — Estimating π

````python
from random import random

TRIES = 10000
hits = 0

for i in range(TRIES):
    x = -1 + 2 * random()
    y = -1 + 2 * random()

    if x * x + y * y <= 1:
        hits = hits + 1

pi_estimate = 4 * hits / TRIES

print("Estimated pi:", pi_estimate)
````

# 4.11 Problem Solving: Solve a Simpler Problem First

## Concept Review

When a full problem feels difficult:

1. solve a smaller version;
2. understand the repeated pattern;
3. generalize the solution;
4. add complexity gradually.

This is especially useful for loop problems because the final solution often repeats a simple operation.

## 4.11.1 Progressive Simplification

Suppose the final task is to process many records.

A useful progression might be:

```text
Process one value
→ process three known values
→ process n known values
→ process values until a sentinel
→ add validation
→ add summary statistics
```

Each step extends a working solution.

## 4.11.2 Worked Example — Sum of Digits

````python
n = 4827
total = 0

while n > 0:
    digit = n % 10
    total = total + digit
    n = n // 10

print("Digit sum:", total)
````

A simpler version of the problem is:

> How do I extract **one** last digit?

Answer:

```python
digit = n % 10
```

Then:

> How do I remove **one** last digit?

Answer:

```python
n = n // 10
```

The loop is simply the repetition of these two simpler operations.
